```python
import numpy as np
import pandas as pd
from scipy.stats import qmc
```

```python
# ------------------------------------------------------------
# Parameter ranges for apple
# ------------------------------------------------------------

parameters = {
     "fruit_albedo":               (0.3, 0.7),
     "fruit_emissivity":           (0.93, 0.98),
     "fruit_diameter_m":           (0.06, 0.10),
     "canopy_resistance_s_m":      (6500.0, 20000.0),
     "diffuse_sw_sky_view":        (0.80, 1.00),
     "longwave_sky_view":          (0.80, 1.00),
     "ground_reflected_sw_view":   (0.00, 0.30),
    "air_temp_bias_C":            (-4.0, 1.0),
     "rh_bias_pct":                (-10.0, 30.0),
     "wind_speed_multiplier":      (0.4, 1.1)
 }


# Parameter names
names = list(parameters.keys())

# Lower and upper bounds
lower_bounds = np.array([parameters[p][0] for p in names])
upper_bounds = np.array([parameters[p][1] for p in names])

# ------------------------------------------------------------
# Number of LHS samples
# ------------------------------------------------------------
N = 500

# ------------------------------------------------------------
# Generate Latin Hypercube Sample
# ------------------------------------------------------------
sampler = qmc.LatinHypercube(
    d=len(names),
    optimization="random-cd",
    seed=42
)

# Samples initially range from 0 to 1
sample_01 = sampler.random(n=N)

# Scale to actual parameter ranges
samples = qmc.scale(
    sample_01,
    lower_bounds,
    upper_bounds
)

# ------------------------------------------------------------
# Convert to dataframe
# ------------------------------------------------------------
lhs_df = pd.DataFrame(
    samples,
    columns=names
)

# Add sample ID
lhs_df.insert(
    0,
    "sample_id",
    np.arange(1, N + 1)
)
```

```python
# ------------------------------------------------------------
# Parameter columns
# ------------------------------------------------------------

parameter_names = [
    "fruit_albedo",
    "fruit_emissivity",
    "fruit_diameter_m",
    "canopy_resistance_s_m",
    "diffuse_sw_sky_view",
    "longwave_sky_view",
    "ground_reflected_sw_view",
    "air_temp_bias_C",
    "rh_bias_pct",
    "wind_speed_multiplier"
]


if "sample_id" not in lhs_df.columns:
    lhs_df.insert(
        0,
        "sample_id",
        np.arange(1, len(lhs_df) + 1)
    )
```

```python

# ------------------------------------------------------------
# Create base runs
# ------------------------------------------------------------
base_runs = lhs_df.copy()

base_runs["base_sample_id"] = base_runs["sample_id"]
base_runs["run_type"] = "base"
base_runs["perturbed_parameter"] = "none"
base_runs["perturbation_fraction"] = 0.0

all_runs = [base_runs]

# ------------------------------------------------------------
# Create +1% perturbation for each parameter
# ------------------------------------------------------------
perturbation = 0.01

for parameter in parameter_names:

    perturbed = lhs_df.copy()

    perturbed["base_sample_id"] = perturbed["sample_id"]

    # Increase only this parameter by 1%
    perturbed[parameter] = (
        perturbed[parameter] * (1 + perturbation)
    )

    # Metadata
    perturbed["run_type"] = "local_sensitivity"
    perturbed["perturbed_parameter"] = parameter
    perturbed["perturbation_fraction"] = perturbation

    all_runs.append(perturbed)

# ------------------------------------------------------------
# Combine everything
# ------------------------------------------------------------
all_runs_df = pd.concat(
    all_runs,
    ignore_index=True
)

# Create unique run ID for HPC runs
all_runs_df.insert(
    0,
    "run_id",
    np.arange(1, len(all_runs_df) + 1)
)
```
